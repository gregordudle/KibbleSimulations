```python
import sympy as sym
import numpy as np
from matplotlib import pyplot as plt
```

```python
point_Bz, B_0 = sym.symbols("point_Bz,B_0")
R, Z, d, g = sym.symbols("R,Z,d,g")     # position Z and R
r,z = sym.symbols("r,z")                # integration variables along the z axis and r axis
```

```python
point_Bz = ((Z - z) / ((R + r)**2 + (Z - z)**2)  - (Z-z)/((R + r + g)**2+(Z-z)**2)) + ((Z - z) / ((-R - r)**2 + (Z - z)**2)  - (Z-z)/((-R - r - g)**2+(Z-z)**2))
display(point_Bz)
B_z = sym.integrate(point_Bz,(z,-d,d))
display(B_z)

#phi = sym.integrate(B_z,(r,1e-6,R))
#display(phi)
```

```python
Z_n = 0.01
R_n = 0.2
d_n = 0.02
g_n = 0.003
specific_B = B_z.subs({(R,R_n),(d,d_n),(g,g_n)})
display(specific_B)
```

```python
r_list = np.arange(1e-4,0.2,1e-3)
print(r_list.size)
B_list = [0] * r_list.size
Z_list = [0.1]
for hoehe in Z_list:
    verlauf = specific_B.subs({(Z,hoehe)})
    i = 0
    for el in r_list:
        B_list[i] = verlauf.subs({(r,el)})
        i +=1
        plt.plot(r_list,B_list)
```
